# Customer Shopping Behavior Analysis

## Project Description

This project focuses on analyzing customer shopping behavior using e-commerce transaction data. 
The analysis explores customer demographics, purchasing patterns, spending behavior, product and 
category preferences, purchase frequency, discount usage, subscription status, shipping preferences, 
payment methods, and review ratings.

The main objective is to understand how different customer characteristics and shopping factors 
are associated with purchasing behavior and to identify meaningful patterns and trends through 
data visualization and exploratory analysis.

The analysis is performed using Python, Pandas, NumPy, Matplotlib, Seaborn, and Plotly.

# Objectives

- Understand the overall profile of customers based on age, gender, and age groups.
- Analyze purchasing and spending patterns across different categories and products.
- Study customer purchase frequency and previous purchase behavior.
- Compare shopping behavior based on discounts and subscription status.
- Analyze customer preferences for shipping and payment methods.
- Explore the relationship between purchase amount, ratings, and other customer-related factors.
- Derive meaningful business insights from the analysis.

# Fill Missing Value with their corresponding Group Median 

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.info()
```

# For any operstion in it we cannot remembering the column name format so we can change their name into lower case and give their sapce'_'

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.columns
```

```python
df = df.rename(columns ={'purchase_amount_(usd)':'purchase_amount'})
```

 # Feature Engineering like create a column age_group
 

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
```

```python
df[['age','age_group']].head(10)
```

# for any operation, column value we change into numerical set

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly':14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

### Correlation Heatmap ⭐⭐⭐

```python
numeric_cols = [
    'age',
    'purchase_amount',
    'review_rating',
    'previous_purchases',
    'purchase_frequency_days'
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(9,6))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap of Customer Behaviour Variables')
plt.show()
```

```python
category_summary = (
    df.groupby('category')
      .agg(
          customers=('customer_id', 'count'),
          avg_purchase=('purchase_amount', 'mean')
      )
      .reset_index()
)

fig = px.bar(
    category_summary,
    x='category',
    y='avg_purchase',
    color='category',
    title='Average Purchase Amount by Category'
)

fig.show()
```
